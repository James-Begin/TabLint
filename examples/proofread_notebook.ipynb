# TabLint in a notebook

Spellcheck for tables, powered by TabPFN-3.5. This notebook opens a **saved** report, so it runs without a GPU or model weights. The last section runs TabPFN-3.5 live on your own DataFrame.

````python
import pandas as pd
import proofread
from proofread import Report

report = Report.load('../results/proofread_demos/breast_injected.json')
report            # rich display: flagged cells underlined, top issues listed
````

## Interactive review
Click a red cell, then **accept** or **dismiss** it (keys: `n`/`p` next/prev, `a` accept, `d` dismiss, `u` undo). Your decisions sync back to Python.

````python
w = report.widget()
w
````

````python
w.decisions.head()        # one row per issue: open / accepted / dismissed
````

````python
cleaned = w.cleaned       # the DataFrame with your accepted fixes applied
cleaned.head()
````
